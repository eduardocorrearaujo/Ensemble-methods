Aplicar a nossa metodologia de ensemble: 

```python
def apply_ensemble(df_preds, weights_lin, weights_log): 

    len_models = len(weights_lin)
    weights_equal =  np.ones(len_models)/len_models

    ens_log = Ensemble(df = df_preds,
        order_models = np.arange(1, 43), 
        dist = 'log_normal',
        mixture = 'log',
        fn_loss = 'median', 
             conf_level=0.95)
  
    ens_lin = Ensemble(df = df_preds,
            order_models = np.arange(1, 43), 
            dist = 'log_normal',
            mixture = 'linear',
            fn_loss = 'median', 
                 conf_level=0.95)
    
    #df_ens_lin = ens_lin.apply_ensemble(weights = weights_lin, p = [0.025, 0.5, 0.975])
    
    df_ens_lin_equal = ens_lin.apply_ensemble(weights = weights_equal, p = [0.025, 0.05, 0.1, 0.25, 0.5, 0.75, 0.9, 0.95, 0.975])

    #df_ens_log = ens_log.apply_ensemble(weights = weights_log, p = [0.025, 0.5, 0.975])
    
    df_ens_log_equal = ens_log.apply_ensemble(weights = weights_equal, p = [0.025, 0.05, 0.1, 0.25, 0.5, 0.75, 0.9, 0.95, 0.975])

    #df_ens_log_crps = ens_log.apply_ensemble(weights = weights_crps, p = [0.025, 0.5, 0.975])
    
    #df_ens_lin_crps = ens_lin.apply_ensemble(weights = weights_crps, p = [0.025, 0.5, 0.975])

    #for df_ in [df_ens_lin, df_ens_lin_equal, df_ens_lin_crps, df_ens_log, df_ens_log_equal, df_ens_log_crps]: 
    #for df_ in [df_ens_lin_equal, df_ens_log_equal]:
    #    df_['step'] = [1,2,3,4]

    return  df_ens_lin_equal, df_ens_log_equal
```

```python
for date_ in df_preds_final.epiweek.unique():
    
    df_preds_ = df_preds_final.loc[df_preds_final.epiweek == date_][['date', 'model_id', 'lower_95', 'pred', 'upper_95']]

    break 
    
df_preds_.head()
```

```python
%%time 
df_ens_lin_end_equal = pd.DataFrame()
df_ens_log_end_equal = pd.DataFrame()

for date_ in df_preds_final.epiweek.unique():

    df_preds_ = df_preds_final.loc[df_preds_final.epiweek == date_][['date', 'model_id', 'lower_95', 'pred', 'upper_95']]

    for date_2 in df_preds_.date.unique(): 

        df_preds_2 = df_preds_.loc[df_preds_.date == date_2]

        len_models = len(df_preds_2.model_id.unique())
    
        weights_equal = np.ones(len_models)/len_models
    
        df_ens_lin_equal, df_ens_log_equal = apply_ensemble(df_preds_2, weights_equal, weights_equal)
    
        df_ens_lin_equal['epiweek'] = date_
    
        df_ens_log_equal['epiweek'] = date_
    
        df_ens_lin_end_equal = pd.concat([df_ens_lin_end_equal, df_ens_lin_equal], ignore_index = True)
        
        df_ens_log_end_equal = pd.concat([df_ens_log_end_equal, df_ens_log_equal], ignore_index = True)
        
```

```python
df_ens_lin_end_equal.to_csv('ensemble_lin_equal.csv', index = False)
df_ens_log_end_equal.to_csv('ensemble_log_equal.csv', index = False)
```
